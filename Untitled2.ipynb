```python
import numpy as np
import matplotlib.pyplot as plt
def func(t,u):
  return 10*np.exp(-(t-2)**2.0/2/(0.075)**2.0-0.6*u)


Tf = 4.0
h = 0.1
N=int(Tf/h)

T= [0]
U=[0.5]
U3=[0.5]

def func(t,u):
  return -5*u

for i in range(0,N):

  k1 = func(T[-1],U[-1])
  k2 = func(T[-1]+h/2,U[-1]+h/2*k1)
  k3 = func(T[-1]+3*h/4,U[-1]+3*h/5*k2)
  k4 = func(T[-1]+h,U[-1]+h/9*(2*k1+3*k2+4*k3))
  U.append(U[-1]+h/9*(2*k1+3*k2+4*k3))
  T.append(T[-1]+h)
  U3.append(U[-1]+h/24*(7*k1+6*k2+8*k3+3*k4))


plt.plot(T,U)
plt.plot(T,U3)
plt.show()
```
